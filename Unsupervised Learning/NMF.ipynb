## Exploring the use of NMF (Non-negative Matrix Factorization) as a dimension reduction technique

```python
## Loading the dataset in csr format
import pandas as pd
from scipy.sparse import csr_matrix

df = pd.read_csv('wikipedia-vectors.csv', index_col=0)
articles = csr_matrix(df.transpose())
titles = list(df.columns)

words = []
# Read words form the file

for line in open('wikipedia-vocabulary-utf8.txt', 'r', encoding='utf-8'):
        word = line.strip()
        words.append(word)
    
    
        
```

```python
words
```

```python
# Import NMF
from sklearn.decomposition import NMF

# Create an NMF instance: model
model = NMF(n_components=6) # Number of topics
# Set n_components to the number of topics you want to extract

# Fit the model to articles
model.fit(articles)

# Transform the articles: nmf_features
nmf_features = model.transform(articles)

# Print the NMF features
print(nmf_features.round(2))
```

### Exploring the constructed components

```python
import pandas as pd

# Create a pandas DataFrame: df
df = pd.DataFrame(nmf_features,index=titles)

# Print the row for 'Anne Hathaway'
print(df.loc['Anne Hathaway'])

# Print the row for 'Denzel Washington'
print(df.loc['Denzel Washington'])
```

#### *NMF components represent topics when applied to a document. NMF features combine topics into documents.
#### For images, NMF components represents patterns or parts of image.

```python
# Import pandas
import pandas as pd

# Create a DataFrame: components_df
components_df = pd.DataFrame(model.components_,columns=words)

# Print the shape of the DataFrame
print(components_df.shape)

# Select row 3: component
component = components_df.iloc[3,:]

# Print result of nlargest
print(component.nlargest())
```

### As saw previously, 3rd feature was most important for Denzel and Anne, and this shows that the topics were related to their field 

### Building Article Recommender System using NMF 

```python
# Perform the necessary imports
import pandas as pd
from sklearn.preprocessing import normalize

# Normalize the NMF features: norm_features
norm_features = normalize(nmf_features)

# Create a DataFrame: df
df = pd.DataFrame(norm_features,index=titles)

# Select the row corresponding to 'Cristiano Ronaldo': article
article = df.loc['Cristiano Ronaldo']

# Compute the dot products: similarities
similarities = df.dot(article)

# Display those with the largest cosine similarity
print(similarities.nlargest())
```

This uses cosine similarity to find closely related topics to Cristiano Ronaldo, and it does find pretty close topics!!

## Recommending Muscial Artists using NMF

```python
# Import artist data 
import numpy as np


artists_array = np.zeros((111, 500), dtype=int)
for line in open('Musical artists/scrobbler-small-sample.csv', 'r', encoding='utf-8'):
    if line.startswith('user_offset'):
        continue  # Skip header line
    col, row, data = line.strip().split(',')
    artists_array[int(row), int(col)] = int(data)

print(artists_array.shape)
print(artists_array)

artists = csr_matrix(artists_array)
print(artists)
df_names = pd.read_csv('Musical artists/artists.csv', header=None)
artist_names = df_names.iloc[:,0].values
```

```python
artists.toarray().shape
print(len(artist_names))
```

```python
# Perform the necessary imports
from sklearn.decomposition import NMF
from sklearn.preprocessing import Normalizer, MaxAbsScaler
from sklearn.pipeline import make_pipeline

# Create a MaxAbsScaler: scaler
scaler = MaxAbsScaler()

# Create an NMF model: nmf
nmf = NMF(n_components=20)

# Create a Normalizer: normalizer
normalizer = Normalizer()

# Create a pipeline: pipeline
pipeline = make_pipeline(scaler,nmf,normalizer)

# Apply fit_transform to artists: norm_features
norm_features = pipeline.fit_transform(artists)
```

### Here we recommend artists most similar to Bruce Springsteen

```python
# Import pandas
import pandas as pd

# Create a DataFrame: df
df = pd.DataFrame(norm_features,index=artist_names)

# Select row of 'Bruce Springsteen': artist
artist = df.loc['Bruce Springsteen']

# Compute cosine similarities: similarities
similarities = df.dot(artist)

# Display those with highest cosine similarity
print(similarities.nlargest())
```
