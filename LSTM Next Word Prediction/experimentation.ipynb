## Next Word Prediction using LSTM RNN


### Data Preprocessing


```python
# Data Collection
import nltk 
from nltk.corpus import gutenberg
nltk.download('gutenberg')
import pandas as pd

# Load the Gutenberg corpus
gutenberg_corpus = gutenberg.raw('shakespeare-hamlet.txt')

with open('gutenberg_corpus.txt', 'w', encoding='utf-8') as f:
    f.write(gutenberg_corpus)


```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split

# Load the text data
with open('gutenberg_corpus.txt', 'r', encoding='utf-8') as f:
    text_data = f.read().lower()

## Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text_data])

total_words = len(tokenizer.word_index) + 1  # +1 for padding token
total_words
```

```python
tokenizer.word_index
```

```python
## Create input sequences
input_sequences = []  # Initialize an empty list to store all input sequences

# Iterate over each line in the text data, splitting by newline character
for line in text_data.split('\n'):
    # Convert the current line into a sequence of integers (tokens) using the tokenizer
    token_list = tokenizer.texts_to_sequences([line])[0]
    
    # Generate n-gram sequences from the tokenized line
    for i in range(1, len(token_list)):
        # For each position in the token list, create a sequence from the start up to the current position (inclusive)
        n_gram_sequence = token_list[:i + 1]
        # Append the n-gram sequence to the list of input sequences
        input_sequences.append(n_gram_sequence)
```

```python
print(f"Total input sequences: {len(input_sequences)}")
```

```python
max_sequence_length = max(len(x) for x in input_sequences)
# Pad sequences: Ensure all sequences have same length, use maximum length sequence from input_sequences
input_sequences = pad_sequences(input_sequences, maxlen=max_sequence_length, padding='pre')
input_sequences
```

```python
import tensorflow as tf
x, y = input_sequences[:,:-1], input_sequences[:,-1] # Last token is the label (y)
print(x)
print(y)
```

```python
y = tf.keras.utils.to_categorical(y, num_classes=total_words) # Convert labels to categorical format
y
```

```python
# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

### Training LSTM RNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

model = Sequential()
model.add(Embedding(input_dim=total_words,output_dim=100, input_length=max_sequence_length-1 )) # Embedding layer
model.add(LSTM(150, return_sequences=True)) 
model.add(Dropout(0.2))
model.add(LSTM(150))
model.add(Dense(total_words, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
model.summary()
```

```python
history = model.fit(x_train, y_train, epochs=100, batch_size=64, validation_data=(x_test, y_test), verbose=1)
```

### PREDICTION

```python
def predict(model,tokenizer, text, max_sequence_length):
    # Convert the input text to lowercase and tokenize it
    text = text.lower()
    token_list = tokenizer.texts_to_sequences([text])[0]

    if len(token_list) > max_sequence_length - 1:
        # If the token list is longer than the max sequence length, truncate it
        token_list = token_list[-(max_sequence_length - 1):]
    
    # Pad the sequence to ensure it matches the model's input shape
    token_list = pad_sequences([token_list], maxlen=max_sequence_length-1, padding='pre')
    
    predicted = model.predict(token_list, verbose=0)
    
    # Get the index of the highest probability word
    predicted_index = np.argmax(predicted, axis=-1)[0]
    
    # Convert the index back to a word
    predicted_word = tokenizer.index_word[predicted_index]
    
    return predicted_word

# Example usage of the predict function
print(predict(model, tokenizer, "To be or not to be considered", max_sequence_length))
```
